```python
### TextLoader

# If needed in a notebook:
# %pip install langchain-community
#from langchain.dcoument_loaders import TextLoader
from langchain_community.document_loaders import TextLoader

loader = TextLoader("../data/text_files/python_intro.txt", encoding="utf-8")
document = loader.load()
print(document)

```

```python
### Directory Loader

from langchain_community.document_loaders import DirectoryLoader

dir_loader = DirectoryLoader(
    "../data/text_files",
    glob="**/*.txt", ## read txt files
    loader_cls= TextLoader, # loader class to use for files
    loader_kwargs={'encoding': 'utf-8'},
    show_progress=True
)

documents = dir_loader.load()

documents
```

```python
### Directory Loader

from langchain_community.document_loaders import PyPDFLoader, PyMuPDFLoader

pdf_dir_loader = DirectoryLoader(
    "../data/pdf",
    glob="**/*.pdf", ## read txt files
    loader_cls= PyMuPDFLoader, # loader class to use for files
    show_progress=True
)

pdf_dcouments = pdf_dir_loader.load()

pdf_dcouments
```
